```python
# =======================================
# NEGACION
# =======================================
# ==== Definicion de la funcion ====
def neg(p) -> bool:
    """Negacion ~p"""
    return False if p else True

# =======================================
# CONJUNCION
# =======================================
# ==== Definicion de la funcion ====
def y(p, q) -> bool:
    """Conjuncion p ^ q"""
    return q if p else False

# =======================================
# DISYUNCION
# =======================================
# ==== Definicion de la funcion ====
def o(p, q) -> bool:
    """Disyuncion p v q"""
    return True if p else q

print("Operadores base (neg, y, o) cargados en memoria.")
```

```python
from inspect import signature

def aridad(fun)-> int:
    """Obtiene la cantidad de parametros de una funcion [cite: 2]"""
    return len(signature(fun).parameters)

def casosPrueba(n: int, res = None) -> list:
    """Genera recursivamente la matriz de 2^n combinaciones booleanas [cite:2]"""
    if res is None:
        res= [[]]
    if n == 0:
        return res
    # Map anade True y False al inicio de cada lista en las combinaciones previas.
    res1 = list(map(lambda t: [True]+t, res))
    res2 = list(map(lambda t: [False]+t, res))

def vectorResultados(Pred)->list:
    """Evalua el predicado en todos sus casos de prueba y retorna la columna final [cite:2]"""
    ar = aridad(Pred)    
    if ar == 0:
        return [Pred()]
    Lvals = casosPrueba(ar)
    return list(map(lambda t: Pred(*t), Lvals))

def logEqv(Pred1,Pred2)-> bool:
        """Compara si dos vectores de resultados son identicos P = Q"""
        return vectorResultados(Pred1) == vectorResultados(Pred2)
```

```python
prompt_ambiguo = lambda p,q,r: o(p,y(q,neg(r)))

# (pvq)^r
prompt_estructurado = lambda p,q,r: y(o(p,q),neg(r))
```
